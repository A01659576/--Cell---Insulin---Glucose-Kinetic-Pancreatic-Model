```python
import sympy as sp

rho, Ge = sp.symbols('rho Ge', positive=True)

R0 = 864; EGO = 1.44; SI = 0.72
sigma = 43.2; alpha = 20000; k = 432
d0 = 0.06; r1 = 0.84e-3; r2 = 0.24e-5

# E1
G1 = (R0 + Ge) / EGO
I1 = 0
beta1 = 0

# E2
G2 = 100
I2 = (R0 + Ge - G2*EGO) / (SI*G2)
beta2 = ((rho + k) * (alpha + G2**2) / (sigma * G2**2)) * ((R0 + Ge - G2*EGO) / (SI * G2))

# E3
G3 = 250
I3 = (R0 + Ge - G3*EGO) / (SI*G3)
beta3 = ((rho + k) * (alpha + G3**2) / (sigma * G3**2)) * ((R0 + Ge - G3*EGO) / (SI * G3))

# Jacobiano evaluado en E1
J1 = sp.Matrix([
    [-(EGO + SI*I1),                          -SI*G1,                         0],
    [2*beta1*sigma*alpha*G1/(alpha+G1**2)**2,  -(rho+k),    sigma*G1**2/(alpha+G1**2)],
    [(r1 - 2*r2*G1)*beta1,                     0,           -d0 + r1*G1 - r2*G1**2]
])


# Jacobiano evaluado en E2
J2 = sp.Matrix([
    [-(EGO + SI*I2),                          -SI*G2,                         0],
    [2*beta2*sigma*alpha*G2/(alpha+G2**2)**2,  -(rho+k),    sigma*G2**2/(alpha+G2**2)],
    [(r1 - 2*r2*G2)*beta2,                     0,           -d0 + r1*G2 - r2*G2**2]
])

# Jacobiano evaluado en E3
J3 = sp.Matrix([
    [-(EGO + SI*I3),                          -SI*G3,                         0],
    [2*beta3*sigma*alpha*G3/(alpha+G3**2)**2,  -(rho+k),    sigma*G3**2/(alpha+G3**2)],
    [(r1 - 2*r2*G3)*beta3,                     0,           -d0 + r1*G3 - r2*G3**2]
])

# Encontrar eigenvalores
eigenvalores_E1 = J1.eigenvals()
eigenvalores_E2 = J2.eigenvals()
eigenvalores_E3 = J3.eigenvals()

det_J1 = sp.simplify(sp.det(J1))
print("det(J1) en E1 =", det_J1)

tr_J1 = sp.simplify(J1.trace())
print("tr(J1) en E1 =", tr_J1)
#  Mostrar el resultado
print(eigenvalores_E1)
print('\n')

det_J2 = sp.simplify(sp.det(J2))
print("det(J2) en E2 =", det_J2)

tr_J2 = sp.simplify(J2.trace())
print("tr(J2) en E2 =", tr_J2)
print(eigenvalores_E2)
print('\n')

det_J3 = sp.simplify(sp.det(J3))
print("det(J2) en E3 =", det_J3)

tr_J3 = sp.simplify(J2.trace())
print("tr(J2) en E3 =", tr_J3)
print(eigenvalores_E3)
print('\n')
```

This first evaluation was to try to find the signs of the eigenvalues by solving the Jacobian matrix. However as it can be seen, the only equilibrium point that can be easily evaluated is the first one, were the results suggest that the equilibrium point is still stable.

However, the results for E1 and E2, are less intuitive, since there are multiple numebers. For that reason, it was decided that it is best to create a Hurwitz-Routh table, where the signs can be evaluated. If the first column of the table only has positive signs, then it's stable, since it indicates that all real parts are negative. In any other case, the equilibrium point is unstable.

```python
from sympy import symbols, Rational, Integer, simplify, Matrix, expand, Poly, fraction

# Ge, p and lambda as symbols
rho, Ge = symbols('rho Ge', positive=True)
lam = symbols('lambda')

# Parámetros
R0=Rational(864); EGO=Rational(144,100); SI=Rational(72,100)
sigma=Rational(432,10); alpha=Integer(20000); k=Integer(432)
d0=Rational(6,100); r1=Rational(84,100000); r2=Rational(24,10000000)


# Equilibrium points from Phase 2
# E1
G1 = (R0 + Ge) / EGO
I1 = Integer(0)
beta1 = Integer(0)

# E2
G2 = Integer(100)
I2 = (R0 + Ge - G2*EGO) / (SI*G2)
beta2 = ((rho + k) * (alpha + G2**2) / (sigma * G2**2)) * ((R0 + Ge - G2*EGO) / (SI * G2))

# E3
G3 = Integer(250)
I3 = (R0 + Ge - G3*EGO) / (SI*G3)
beta3 = ((rho + k) * (alpha + G3**2) / (sigma * G3**2)) * ((R0 + Ge - G3*EGO) / (SI * G3))


def Jacobian(G, I, b):
    return Matrix([
        [-(EGO+SI*I),                      -SI*G,                    0],
        [2*b*sigma*alpha*G/(alpha+G**2)**2, -(rho+k), sigma*G**2/(alpha+G**2)],
        [(r1-2*r2*G)*b,                     0,        -d0+r1*G-r2*G**2]
    ])

# In this case, the polynomial is cubic because the matrix is 3X3

def sign_asses(expr):
    """Determine if the expression is positive or negative.
       Routh - Hurwitz. Consider the cubic polynomial
       𝜆^3 + a2𝜆^2 + a1𝜆 + a0
       Hurwitz table would be
       a3                   a1
       a2                   a0
      (a2a1 - a3a0)/a2      0
       a0
    """
    #  simplify the expression and split denuminator with numerator to assess signs
    num, den = fraction(simplify(expr))
    # extract coefficients from polynomials of numinator and denumerator
    num_coeffs = Poly(expand(num), Ge, rho).coeffs()
    den_coeffs = Poly(expand(den), Ge, rho).coeffs()
    # Verify the signs of the coefficients
    num_pos = all(c > 0 for c in num_coeffs)
    num_neg = all(c < 0 for c in num_coeffs)
    den_pos = all(c > 0 for c in den_coeffs)
    den_neg = all(c < 0 for c in den_coeffs)
    # +/+ or -/- -> positive
    if (num_pos and den_pos) or (num_neg and den_neg):
        return "positive"
    # +/- or -/+ -> -
    elif (num_pos and den_neg) or (num_neg and den_pos):
        return "negative"
    else:
        return "undetermined"

def analysis(G_val, I_val, b_val):
    print("="*60)

    # Get Jacobian por each equilibrium point
    J  = Jacobian(G_val, I_val, b_val)
    # Calculate de characteristic polynomial
    p  = J.charpoly(lam).as_expr()

    # extract coefficients in descendent order and simplify
    _, a2, a1, a0 = [simplify(c) for c in Poly(expand(p), lam).all_coeffs()]
    # Calculate b1 and c1
    # In this case a3 = 1, since it's the Jacobian is 3X3
    b1 = simplify(a2*a1 - a0) / simplify(a2)
    c1 = a0 # last row in Hurwitz is a0.

    # print signs of the column in Hurwitz table
    print(f"\n  Signs for all Ge, rho > 0:")
    s_a2 = sign_asses(a2); s_b1 = sign_asses(b1); s_c1 = sign_asses(c1)
    print(f"  a2 is {s_a2}")
    print(f"  b1 is {s_b1}")
    print(f"  c1 is {s_c1}")
    print('\n')

    print(f'a1: {a1}')
    print(f'b1: {b1}')
    print(f'c1: {c1}')

    # if all signs are positive, then the real part is negative.
    # any other case, it's unstable
    if all("positive" in s for s in [s_a2, s_b1, s_c1]):
        print(f"\n  All eigenvalues have negative real parts -> Stable")
    else:
        print(f"\n   The equilibrium is unstable")

print('Point of Equilibrium 1')
analysis(G1, I1, beta1)
print('Point of Equilibrium 2')
analysis(G2, I2, beta2)
print('Point of Equilibrium 3')
analysis(G3, I3, beta3)
```

The results from the Hurwitz-Routh table, shows how the behaviour of the equilibrium points remain the same, regardless of the free values $Ge$ and $\rho$. Hence, the first and second equilibrium points, stay stable, while the third one is unstable.

For this case, we assess the term $$
\frac{d\beta}{dt} = \left(-d_0 + r_1 G - r_2 G^2\right)\beta
$$.

The first possibility is that $β = 0$, in that case $r_1$ nor $r_2$ would have an effect on the system.

However, when $β \ne 0$, then the term that has to be considered is $(-d_0 + r_1 G - r_2 G^2)$

During fase 2, the roots of these polynomial showed that there are two solutions $G_1 = 100$ and $G_2 = 250$. However, in this case the discrimintant of the roots will determine the effect of havind these two parameters as free.

The discriminant in this case is: $r_1^2 - 4r_2d_0$.

There are three possible cases.


*   **Case 1:** $r_1^2 < 4r_2d_0$

  In this case, there would be no real solutions, since the discrimintant would be negative, hence there would be imaginary results. This means that the parabola will never change directions. In particular, we observe that the direction of the term will be negative since $(-d_0 + r_1 G - r_2 G^2)$, has a concave down parabola. Given that in this case the term decreades and that $β > 0$, then the derivate is negative as well.

  Since, $β$, decreses, then it will eventually converge to zero, leading to the first equilibrium point, which is stable.

*   **Case 2:** $r_1^2 = 4r_2d_0$

  For this case G = ${2}\frac {-r_1}{-r_2}  = 2\frac{r_1}{r_2}$


 This means that the parabola only crosses one point. As soon as the system moves, this will continue to be negative, since there is no change of signs. In the case where G is exactly equal to zero, then the derivative of $β$ would be zero, meaning that the amount of cells remain constant at all times.

  *   **Case 3:** $r_1^2 > 4r_2d_0$

  This case, leads to two positive solutions, since all the terms are positive. Hence, in this case the derivative of $β$ will increase.
